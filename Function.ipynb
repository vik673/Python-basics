```python
rev_list = []
def reverse_list(sample_list):
    for reverse_item in sample_list[::-1]:
        rev_list.append(reverse_item)
    return rev_list
```

```python
def sum_of_numbers(num_param):
    even_sum = 0
    odd_sum = 0
    for num in range(num_param):
        if(num % 2 == 0):
            even_sum += num
        else:
            odd_sum +=num
    lst = [even_sum, odd_sum]
    return lst
```

```python
print(sum_of_numbers(10))
```

```python
def evens_and_odds(num_param):
    even_count = 0
    odd_count = 0
    for num in range(num_param+1):
        if(num % 2 == 0):
            even_count = even_count + 1
        else:
            odd_count = odd_count + 1
    lst = [even_count, odd_count]
    return lst
```

```python
print(evens_and_odds(100))
```

```python
def factorial(number):
    fact = 1
    for num in range(1, number+1):
        fact *= num
    return fact

print(factorial(5))
```

```python
def is_empty(param):
    if param:
        return False
    else:
        return True
        
print(is_empty([]))        
print(is_empty([1, 2, 3])) 
print(is_empty(""))        
print(is_empty("Hello"))   
print(is_empty({}))      
print(is_empty({"key": "value"})) 
```

```python
import math

def is_prime(num):
    str = ''
    flag = False
    r1 = int(math.sqrt(num))
    for n1 in range(2,r1+1):
        if(num%n1 == 0):
            flag = True
            break
    if(flag):
         str = "Number is not prime"
    else:
        str = "Number is prime"
    return str

print(is_prime(7))
```
